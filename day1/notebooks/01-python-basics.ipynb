## 3. Dictionaries: data with labels

```python
robot = {"name": "Vikky", "battery": 80, "skills": ["talk", "see", "listen"]}
print(robot["name"], "can", ", ".join(robot["skills"]))
robot["battery"] = robot["battery"] - 65
print("Battery now:", robot["battery"])
```

## 5. Functions: name a job, reuse it

```python
def greet(name, style="friendly"):
    if style == "pirate":
        return f"Ahoy, {name}!"
    return f"Hi {name}, nice to meet you."

print(greet("Kuntal"))
print(greet("Meena", style="pirate"))
```

## 6. This is what an AI call is made of
A list of dicts. Each has a `role` and some `content`. Notebook 02 sends exactly this to Vikky.

```python
messages = [
    {"role": "system", "content": "You are a friendly robot tutor."},
    {"role": "user", "content": "What is a sensor?"},
]
for m in messages:
    print(m["role"], "->", m["content"])
```
